```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import roc_auc_score
from xgboost import XGBClassifier

# 1. Define columns
categorical_col = X_train.select_dtypes(exclude=['number']).columns.tolist()
numeric_col = X_train.select_dtypes(include=['number']).columns.tolist()

# 2. Preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_col),
        ('num', StandardScaler(), numeric_col)
    ]
)

# 3. Pipeline: preprocessing + model
pipe = Pipeline([
    ('preprocessor', preprocessor),
    ('model', XGBClassifier(
        random_state=42,
        eval_metric='logloss'
    ))
])

# 4. Parameter search space
param_dist = {
    'model__learning_rate': [0.05, 0.10, 0.15, 0.20, 0.25, 0.30],
    'model__max_depth': [3, 4, 5, 6, 8, 10, 12, 15],
    'model__min_child_weight': [1, 3, 5, 7],
    'model__gamma': [0.0, 0.1, 0.2, 0.3, 0.4],
    'model__colsample_bytree': [0.3, 0.4, 0.5, 0.7]
}

# 5. Random search
rs_model = RandomizedSearchCV(
    estimator=pipe,
    param_distributions=param_dist,
    n_iter=5,
    scoring='roc_auc',
    n_jobs=-1,
    cv=5,
    verbose=3,
    random_state=42
)

# 6. Fit on original training data
rs_model.fit(X_train, y_train)

# 7. Best model and best CV score
print("Best Params:", rs_model.best_params_)
print("Best CV ROC-AUC:", rs_model.best_score_)

# 8. Test-set evaluation
best_model = rs_model.best_estimator_

y_pred_tuned = best_model.predict_proba(X_test)
y_prob_tuned = y_pred_tuned[:, 1]

roc_auc_tuned = roc_auc_score(y_test, y_prob_tuned)
print("Test ROC-AUC:", roc_auc_tuned)
```
